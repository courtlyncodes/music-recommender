```python
import pandas as pd
import numpy as np
```

```python
songs = pd.DataFrame({
    "title": [
        "Song A",
        "Song B",
        "Song C",
        "Song D",
        "Song E"
    ],
    "artist": [
        "Artist 1",
        "Artist 2",
        "Artist 3",
        "Artist 4",
        "Artist 5"
    ],
    "danceability": [0.90, 0.82, 0.35, 0.72, 0.25],
    "energy": [0.85, 0.75, 0.30, 0.65, 0.20],
    "popularity": [90, 75, 40, 65, 30]
})

songs
```

```python
features = songs[["danceability", "energy"]]

features
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from sklearn.metrics.pairwise import euclidean_distances
```

```python
distance_matrix = euclidean_distances(features)

distance_matrix
```

```python
def recommend_songs(song_title, n=3):
    song_index = songs[songs["title"] == song_title].index[0]

    distances = distance_matrix[song_index]
    ranked_indexes = np.argsort(distances)

    recommendations = ranked_indexes[1:n + 1]

    return songs.iloc[recommendations][["title", "artist"]]
```

```python
recommend_songs("Song A")
```
